## Model Benchmark Exploration

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "model": ["Linear Reg", "Random Forest", "Neural Net", "XGBoost"],
    "accuracy": [0.72, 0.89, 0.94, 0.91]
})

#Display the table
print(df)

#Display an inline plot
plt.figure(figsize=(6, 3))
plt.bar(df["model"], df["accuracy"], color='skyblue')
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")
plt.show()
```
